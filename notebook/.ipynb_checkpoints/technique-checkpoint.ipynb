# DONNEES

Les données MICS Togo 2017 correspondent à la sixième édition de l'enquête Multiple Indicator Cluster Survey (MICS6), réalisée au Togo en 2017 par l'INSEED, avec l'appui technique de l'UNICEF.

**Quelques informations techniques** : 
- Identifiant de l'enquête : TGO_2017_MICS_v01_M
- Pays : Togo
- Année : 2017
- Producteur : INSEED
- Appui technique : UNICEF
- Format des microdonnées : SPSS (.sav)

# CONTEXTE & OBJECTIFS

### Probleme
Mesurer la taille et le poids de chaque enfant est lent et coûteux. Les équipes de terrain(ONG, agents de santé) ont des moyens limités : elles ne peuvent pas tout mesurer partout en même temps. Remplir un questionnaire sur le ménage est, au contraire, rapide.

### Idee
Utiliser des réponses de questionnaire (région, âge de l'enfant, toilettes, électricité, instruction de la mère...) pour estimer, sans balance ni toise, le risque de malnutrition de chaque enfant, puis du ménage.

### Ce que l'outil est, et n'est pas
L'outil classe, il ne diagnostique pas. Un ménage "à risque élevé" est un ménage àvisiter en premier, pas un ménage où l'on sait qu'un enfant est malade. **Il ne remplace jamais la mesure**.

   - Stunting : 3678(75.62%) enfants (0 : non-atteints) | 1186(24.38%) enfants (1 : atteints)
 - Resultat sur les niveaux de risque des enfants atteints (1 : MOYEN | 2 : ELEVEE) :
    - Wasting_niveau : 1 ---- 222 enfants ---- 79.57% ||2 ---- 57 enfants ---- 20.43%
    - Stunting_niveau : 1 ---- 834 enfants ---- 70.32% || 2 ---- 352 enfants ---- 29.68%
